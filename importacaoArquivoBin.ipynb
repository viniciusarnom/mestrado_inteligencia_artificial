```python
import numpy as np
import json
import matplotlib.pyplot as plt

# -----------------------------
# Caminhos dos arquivos
# -----------------------------
bin_file = "WiFi_Day_1_bes_s1.bin"
json_file = "WiFi_Day_1_bes_s1.json"

# -----------------------------
# Carregar metadados do JSON
# -----------------------------
with open(json_file, "r") as f:
    metadata = json.load(f)

datatype = metadata["global"]["core:datatype"]
sample_rate = int(metadata["global"]["core:sample_rate"])
num_samples = int(metadata["annotations"]["core:sample_count"])

print("Datatype:", datatype)
print("Sample rate:", sample_rate)
print("Número de amostras:", num_samples)

# -----------------------------
# Ler arquivo BIN
# -----------------------------
# cf32 = complex float32
iq_samples = np.fromfile(bin_file, dtype=np.complex64)

print("Total de amostras lidas:", len(iq_samples))

# -----------------------------
# Separar componentes I e Q
# -----------------------------
I = iq_samples.real
Q = iq_samples.imag

print("Primeiras amostras:")
print(iq_samples[:10])

# -----------------------------
# Plot simples do sinal
# -----------------------------
plt.figure()
plt.plot(I[:200], label="I")
plt.plot(Q[:200], label="Q")
plt.legend()
plt.title("Amostras IQ")
plt.show()
```
