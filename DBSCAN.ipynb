```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar o dataset
df = pd.read_csv('dataset_balanceado.csv')  # Ajuste o caminho conforme necessário

# Separar atributos (X) e rótulo (y)
X = df.drop(columns=['continente'])
y = df['continente']

# Codificar colunas categóricas em X
for col in X.select_dtypes(include='object').columns:
    X[col] = LabelEncoder().fit_transform(X[col])

# Codificar rótulo
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Normalizar os dados
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aplicar DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)  # ajuste o eps e min_samples conforme necessário
clusters = dbscan.fit_predict(X_scaled)

# Avaliar número de clusters
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
n_noise = list(clusters).count(-1)

# Avaliação da qualidade da clusterização
if n_clusters > 1:
    silhouette = silhouette_score(X_scaled, clusters)
    ari = adjusted_rand_score(y_encoded, clusters)
    print(f"Silhouette Score: {silhouette:.4f}")
    print(f"Adjusted Rand Index (comparado com 'continente'): {ari:.4f}")
else:
    print("Menos de 2 clusters identificados. Silhouette Score não pode ser calculado.")

print(f"Número de clusters encontrados: {n_clusters}")
print(f"Número de pontos classificados como ruído: {n_noise}")
```
