```python
import numpy as np
import matplotlib.pyplot as plt

# arquivos
file_day1 = "WiFi_Day_1_bes_s1.bin"
file_day2 = "WiFi_Day_2_bes_s1.bin"

# taxa de amostragem do JSON (5 MHz)
fs = 5e6


def load_iq(file):

    iq = np.fromfile(file, dtype=np.complex64)

    I = iq.real
    Q = iq.imag

    return I, Q


# carregar sinais
I1, Q1 = load_iq(file_day1)
I2, Q2 = load_iq(file_day2)

# pegar apenas uma janela pequena para visualização
N = 5000

I1 = I1[:N]
Q1 = Q1[:N]

I2 = I2[:N]
Q2 = Q2[:N]

# eixo do tempo
t = np.arange(N) / fs


plt.figure(figsize=(12,6))

# Day 1
plt.subplot(2,1,1)

plt.plot(t, I1, label="I")
plt.plot(t, Q1, label="Q", alpha=0.7)

plt.title("Sinal IQ no tempo - Day 1")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.legend()


# Day 2
plt.subplot(2,1,2)

plt.plot(t, I2, label="I")
plt.plot(t, Q2, label="Q", alpha=0.7)

plt.title("Sinal IQ no tempo - Day 2")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.legend()


plt.tight_layout()
plt.show()
```
