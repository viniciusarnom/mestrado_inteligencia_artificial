```python
import numpy as np
import matplotlib.pyplot as plt

# Ler os primeiros 5000 samples (cf32 = complex64 no numpy)
# cf32 significa float32 complexo (32 bits para I, 32 bits para Q)
data = np.fromfile('WiFi_Day_1_bes_s1.bin', dtype=np.complex64, count=5000)

I = np.real(data)
Q = np.imag(data)

# Criar o gráfico
plt.figure(figsize=(14, 6))
plt.plot(I[:1000], label='Parte I (In-phase)', color='blue')
plt.plot(Q[:1000], label='Parte Q (Quadrature)', color='red', alpha=0.7)
plt.title('Gráfico das Partes I e Q do Sinal (Primeiras 1000 amostras)')
plt.xlabel('Índice da Amostra')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('iq_components_plot.png')
```
